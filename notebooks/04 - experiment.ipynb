```python
import json, os

import sys
sys.path.append("..")


import src.config as config
import src.models.datacenter as datacenter
import src.models.requests as requests
import src.models.orchestrator as orchestrator
```

```python
# Load data centers and requests

request_queue= requests.RequestQueue()
datacenters_data_folder = f"{config.experiments_folder}/{config.d_i}-{config.d_f}/processed/"
datacenters = {}
for dc_name in os.listdir(datacenters_data_folder):
    datacenters[dc_name] = datacenter.Datacenter(
        name=dc_name,
        data=json.load(open(datacenters_data_folder+dc_name+"/profile.json", "r"))
    )
    request_queue.load(datacenters_data_folder+dc_name+"/requests.csv", dc_name.split("_")[1])

```

```python
import src.models.algorithms as algorithms
methods = {
    "geo_based": algorithms.geo_based,
    "carbon_greedy":algorithms.carbon_greedy,
    "water_greedy": algorithms.water_greedy,
    "land_use_greedy": algorithms.land_use_greedy

}
```

```python
out_dir = f"{config.output_folder}/{config.d_i}-{config.d_f}/results"
import os 
if not os.path.exists(out_dir):
    os.makedirs(out_dir) 

for algorithm_name, algorithm in methods.items():
    orch = orchestrator.Orchestrator(
        jobs=request_queue.requests_list,
        datacenters=datacenters
    )

    orch.run_simulation(scheduling_function = algorithm)

    with open(f"{out_dir}/{algorithm_name}.json", "w") as f:  
        json.dump({
            "traces": [job.trace for job in orch.jobs]
            }, f, indent=4)
```
