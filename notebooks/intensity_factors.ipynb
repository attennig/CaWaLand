```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import numpy as np

# Data from the image
factors = ['CI', 'EWIF', 'ELIF']
sources = ['wind', 'nuclear', 'hydro', 'geothermal', 'solar', 'unknown', 'biomass', 'gas', 'oil', 'coal']
```

```python
data = []
for factor, coefficients in parameters.coefficients_raw.items():
    print(factor)
    values = []
    for source in sources:
        print(f"  {source}: {coefficients[source]}")
        min_val = min(coefficients.values())
        max_val = max(coefficients.values())
        denom = max_val - min_val
        minmax_norm = 0.0 if denom == 0 else (coefficients[source] - min_val) / denom
        sum_norm = coefficients[source] / sum(coefficients.values())
        print(f"  min-max normalized: {minmax_norm}")
        print(f"  sum normalized: {sum_norm}")
        values.append(sum_norm)
    data.append(values)
```

```python

data = np.array(data)

fig, ax = plt.subplots(figsize=(12, 4), dpi=300)

# Custom colormap: light yellow to pink/red
cmap = mcolors.LinearSegmentedColormap.from_list('custom', ['#f9f0a8', '#f4a582', '#d6604d'], N=256)

im = ax.imshow(data, cmap=cmap, aspect='auto', vmin=0, vmax=0.7)

# Colorbar
cbar = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.02)
cbar.set_label('Intensity', fontsize=11)
cbar.ax.tick_params(labelsize=9)

# Ticks
ax.set_xticks(range(len(sources)))
ax.set_xticklabels(sources, rotation=45, ha='right', fontsize=10)
ax.set_yticks(range(len(factors)))
ax.set_yticklabels(factors, fontsize=11, fontweight='bold')

# Labels
ax.set_xlabel('Energy Source', fontsize=12, labelpad=10)
ax.set_ylabel('Factor', fontsize=12, labelpad=10)
ax.set_title('Normalized Intensity Factors by Energy Source (ordered by carbon)',
             fontsize=13, fontweight='bold', pad=14)

# Annotate cells
for i in range(len(factors)):
    for j in range(len(sources)):
        val = data[i, j]
        # Format value
        if val == 0:
            text = '0'
        elif val < 0.001:
            text = f'{val:.5f}'
        elif val < 0.01:
            text = f'{val:.4f}'
        elif val < 0.1:
            text = f'{val:.3f}'
        else:
            text = f'{val:.2g}'
        # Choose text color based on background intensity
        color = 'black' if val < 0.4 else 'white'
        ax.text(j, i, text, ha='center', va='center', fontsize=8.5,
                color=color, fontweight='medium')

plt.tight_layout()
plt.savefig('./experiments/plots/intensity_factors.pdf', format='pdf', bbox_inches='tight', dpi=300)
print("Saved as intensity_factors.pdf")
```
